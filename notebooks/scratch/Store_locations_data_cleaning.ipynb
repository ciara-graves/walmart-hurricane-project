```python
import pandas as pd
import re

stores_raw

rows_before = len(stores_raw)


def clean_store_locations(df):
    clean = df.copy()

    text_columns = [
        "store_id", "store_name", "city",
        "state", "region"
    ]

    for col in text_columns:
        clean[col] = clean[col].astype("string").str.strip()

    clean["store_id"] = clean["store_id"].str.upper()
    clean["store_id"] = clean["store_id"].str.replace(
        r"^([A-Z]{2})[-_]?(\d+)$",
        r"\1-\2",
        regex=True
    )

    state_mapping = {
        "FLORIDA": "FL",
        "FLA.": "FL",
        "FL": "FL"
    }

    clean["state"] = (
        clean["state"]
        .str.upper()
        .replace(state_mapping)
    )

    clean["city"] = clean["city"].str.title()
    clean["store_name"] = clean["store_name"].str.title()

    clean["region"] = clean["region"].str.strip()

    clean = clean.drop_duplicates()

    clean = clean.drop_duplicates(
        subset=["store_id"],
        keep="first"
    )

    return clean


stores_clean = clean_store_locations(stores_raw)

rows_after = len(stores_clean)

print("Rows before cleaning:", rows_before)
print("Rows after cleaning:", rows_after)
print("Rows removed:", rows_before - rows_after)

display(stores_clean)

print("\nMissing values:")
print(stores_clean.isna().sum())

print("\nDuplicate store IDs:", stores_clean["store_id"].duplicated().sum())

print("\nData types:")
print(stores_clean.dtypes)
```
